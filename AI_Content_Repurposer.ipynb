```python

import gradio as gr
from pypdf import PdfReader
from PIL import Image
import pytesseract
import cv2
import tempfile
import os


def extract_content(pdf_file, image_file, video_file, input_text):
    content_parts = []

    # PDF text
    if pdf_file:
        try:
            reader = PdfReader(pdf_file)
            pdf_text = "\n".join(
                page.extract_text() or "" for page in reader.pages
            )
            if pdf_text.strip():
                content_parts.append("PDF CONTENT:\n" + pdf_text)
        except Exception as e:
            content_parts.append(f"PDF could not be read: {e}")

    # Photo: extract visible text using OCR
    if image_file:
        try:
            img = Image.open(image_file).convert("RGB")
            image_text = pytesseract.image_to_string(img)
            if image_text.strip():
                content_parts.append("TEXT FOUND IN PHOTO:\n" + image_text)
            else:
                content_parts.append(
                    "A photo was uploaded, but no readable text was detected. "
                    "This text-only model cannot describe the image itself."
                )
        except Exception as e:
            content_parts.append(f"Photo processing failed: {e}")

    # Video: sample frames and extract visible text
    if video_file:
        cap = cv2.VideoCapture(video_file)
        frame_texts = []
        frame_count = 0

        try:
            total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
            fps = cap.get(cv2.CAP_PROP_FPS)

            if fps and total_frames:
                duration = total_frames / fps
                sample_times = [0, duration / 2, max(0, duration - 1)]

                for seconds in sample_times:
                    cap.set(cv2.CAP_PROP_POS_MSEC, seconds * 1000)
                    ok, frame = cap.read()

                    if ok:
                        frame_count += 1
                        frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
                        frame_img = Image.fromarray(frame)
                        found = pytesseract.image_to_string(frame_img)

                        if found.strip():
                            frame_texts.append(found.strip())

            if frame_texts:
                content_parts.append(
                    "TEXT FOUND IN SAMPLED VIDEO FRAMES:\n"
                    + "\n".join(frame_texts)
                )
            else:
                content_parts.append(
                    "Video uploaded. No readable text was found in the "
                    "sampled frames. Video audio and scene meaning are "
                    "not analyzed in this version."
                )
        except Exception as e:
            content_parts.append(f"Video processing failed: {e}")
        finally:
            cap.release()

    # Pasted text
    if input_text and input_text.strip():
        content_parts.append("USER TEXT:\n" + input_text.strip())

    combined = "\n\n".join(content_parts).strip()
    return combined[:5000]


def generate_content(pdf_file, image_file, video_file,
                     input_text, content_type, language):

    content = extract_content(
        pdf_file, image_file, video_file, input_text
    )

    if not content:
        return "Please upload a file or paste some text.", None

    if "no readable text" in content.lower() and len(content) < 250:
        return (
            content + "\n\nPlease also paste a description or transcript "
            "so the text model has content to work with.",
            None
        )

    formats = {
        "Summary": (
            "Summarize the main ideas in 3 to 5 clear sentences."
        ),
        "Instagram Caption": (
            "Write one engaging caption of 2 to 3 sentences, followed "
            "by up to 5 unique, relevant hashtags."
        ),
        "LinkedIn Post": (
            "Write a professional opening, a short body, and a takeaway."
        ),
        "Reel Script": (
            "Write HOOK, NARRATION, CLOSING, and ON-SCREEN TEXT sections. "
            "Make it suitable for a 30-second video."
        ),
        "Blog Post": (
            "Write a title, short introduction, two body paragraphs, "
            "and a conclusion."
        ),
        "Podcast Script": (
            "Write an INTRO, MAIN DISCUSSION, and CONCLUSION."
        ),
        "Headlines": (
            "Generate exactly five different headlines in a numbered list."
        ),
        "Hashtags": (
            "Generate up to eight relevant, unique hashtags."
        )
    }

    prompt = f"""<|im_start|>system
You are a content creation assistant.
Follow the requested structure.
Use only information supported by the supplied source.
Avoid repetition. Write in {language}.
<|im_end|>
<|im_start|>user
TASK: {content_type}
INSTRUCTIONS: {formats[content_type]}

SOURCE:
{content}

Write the finished output now.
<|im_end|>
<|im_start|>assistant
"""

    try:
        result = summarizer(
            prompt,
            max_new_tokens=180,
            do_sample=False,
            return_full_text=False,
            repetition_penalty=1.15
        )

        generated = result[0]["generated_text"].strip()

        with tempfile.NamedTemporaryFile(
            mode="w",
            suffix=".txt",
            prefix="generated_content_",
            encoding="utf-8",
            delete=False
        ) as temp:
            temp.write(generated)
            output_path = temp.name

        return generated, output_path

    except Exception as e:
        return f"Generation failed: {e}", None


with gr.Blocks(
    title="AI Content Studio",
    theme=gr.themes.Soft()
) as demo:

    gr.Markdown("""
    # ✨ AI Content Studio
    ### Turn your content into multiple formats
    Upload a PDF, photo, video, or paste text.
    """)

    with gr.Row():
        with gr.Column():
            pdf_input = gr.File(
                label="Upload PDF",
                file_types=[".pdf"],
                type="filepath"
            )

            image_input = gr.File(
                label="Upload Photo (JPG or PNG)",
                file_types=[".jpg", ".jpeg", ".png"],
                type="filepath"
            )

            video_input = gr.File(
                label="Upload Video (MP4, MOV, or AVI)",
                file_types=[".mp4", ".mov", ".avi"],
                type="filepath"
            )

            text_input = gr.Textbox(
                label="Or paste text / transcript",
                placeholder="Paste your article, notes, or transcript...",
                lines=6
            )

            content_type = gr.Dropdown(
                choices=[
                    "Summary",
                    "Instagram Caption",
                    "LinkedIn Post",
                    "Reel Script",
                    "Blog Post",
                    "Podcast Script",
                    "Headlines",
                    "Hashtags"
                ],
                value="Reel Script",
                label="Choose output type"
            )

            language = gr.Dropdown(
                choices=[
                    "English", "Tamil", "Hindi",
                    "Kannada", "Telugu", "Malayalam"
                ],
                value="English",
                label="Output language"
            )

            generate_button = gr.Button(
                "✨ Generate Content",
                variant="primary"
            )

        with gr.Column():
            output_text = gr.Textbox(
                label="Generated Content",
                lines=16,
                interactive=False
            )

            download_file = gr.File(
                label="Download generated content"
            )

    generate_button.click(
        fn=generate_content,
        inputs=[
            pdf_input, image_input, video_input,
            text_input, content_type, language
        ],
        outputs=[output_text, download_file]
    )

demo.launch(share=True)
```
